```python
import pandas as pd
import os

# Define paths to data files
RAW_DATA_DIR = os.path.join("..","data","raw")

PATH_FTE_ACTUAL = os.path.join(RAW_DATA_DIR, "fte-enrollments-actual.csv")
PATH_FTE_TARGET = os.path.join(RAW_DATA_DIR, "fte-enrollments-target.csv")
PATH_GRANTS = os.path.join(RAW_DATA_DIR, "operating-grants.csv")

# Load the datasets
df_fte_actual = pd.read_csv(PATH_FTE_ACTUAL)
df_fte_target = pd.read_csv(PATH_FTE_TARGET)
df_grants = pd.read_csv(PATH_GRANTS)
```

```python
# Inspection of fte_actual dataset
df_fte_actual.info()
display(df_fte_actual.head())
display(df_fte_actual.describe())
```

```python
# Inspection of fte_target dataset
df_fte_target.info()
display(df_fte_target.head())
display(df_fte_target.describe())
```

```python
# Inspection of grants dataset
df_grants.info()
display(df_grants.head())
display(df_grants.describe())
```

```python
# All datasets have 26 rows. Check whether the institutions are the samme across datasets
actual_insts = set(df_fte_actual['Institution'].dropna().unique())
target_insts = set(df_fte_target['Institution'].dropna().unique())
grants_insts = set(df_grants['Institution'].dropna().unique())

# Identify mismatches between target and actual
print("In Target but not Actual:", target_insts - actual_insts)
print("In Actual but not Target:", actual_insts - target_insts)

# Identify mismatches grants and actual
print("In Grants but not Actual:", grants_insts - actual_insts)
print("In Actual but not Grants:", actual_insts - grants_insts)
```

```python
# Verify that string stripping resolves target vs actual discrepancies
actual_clean = set(df_fte_actual['Institution'].str.strip().unique())
target_clean = set(df_fte_target['Institution'].str.strip().unique())
grants_clean = set(df_grants['Institution'].str.strip().unique())

unmatched_target = actual_clean ^ target_clean
unmatched_grants = actual_clean ^ grants_clean

print(f"Unmatched Target institutions after strip: {len(unmatched_target)}")
print(f"Unmatched Grants institutions after strip: {len(unmatched_grants)}")

if len(unmatched_target) == 0 and len(unmatched_grants) == 0:
    print("`.str.strip()` resolves all institution name mismatches across datasets")
else:
    print(f"Remaining discrepancies: {unmatched_target | unmatched_grants}")
```

```python
# Examine fiscal year coverage across datasets
actual_years = set(df_fte_actual['Fiscal Year'].dropna().unique())
target_years = set(df_fte_target['Fiscal Year'].dropna().unique())
grants_years = set(df_grants['Fiscal Year'].dropna().unique())

# Identify fiscal year overlap and boundary differences
common_years = actual_years & target_years & grants_years
all_years = actual_years | target_years | grants_years

print(f"Total distinct fiscal years across all files: {len(all_years)}")
print(f"Fiscal years common to all three datasets: {len(common_years)} years ({min(common_years)} to {max(common_years)})")
print("Years in Actual but missing in Target:", sorted(list(actual_years - target_years)))
print("Years in Target but missing in Actual:", sorted(list(target_years - actual_years)))
```
